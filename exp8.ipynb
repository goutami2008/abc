```python
import pandas as pd
```

```python
df=pd.read_csv("CreditCard_DV (10).csv")
df
```

```python
import matplotlib.pyplot as plt
```

```python
fig=plt.figure(figsize=[5,5])
ax=fig.add_axes([0,0,1,1])
ax.scatter(df['Credit_Limit'],df['Avg_Purchase'])
ax.set_title("Credit_Limit vs Avg_Purchase",fontsize=16)
ax.set_xlabel('Credit_Limit',fontsize=12)
ax.set_ylabel('Avg_Purchase',fontsize=12)
fig.show()
```

```python
import matplotlib.pyplot as plt
```

```python
df_list=(df['Credit_Limit'],df['Avg_Purchase'])
fig=plt.figure(figsize=[6,5])
ax=fig.add_axes([0,0,1,1])
ax.set_title("Credit_Limit" and "Avg_Purchase")
ax.boxplot(df_list,widths=0.5)
plt.show()
```

```python
grouped_df=df.groupby('Income_Category').size().reset_index(name="count")
fig=plt.figure(figsize=[7,5])
ax=fig.add_axes([0,0,1,1])
ax.set_title("total number of income category",fontsize=16)
ax.bar(grouped_df["Income_Category"],grouped_df["count"])
ax.set_xlabel("Income_Category",fontsize=12)
ax.set_ylabel("count",fontsize=12)
plt.tight_layout()
plt.show()
```

```python
fig=plt.figure(figsize=[7,5])
ax=fig.add_axes([0,0,1,1])
ax.hist(df['Total_Trans_Amt'],bins=20,edgecolor='black')
ax.set_title("distrubution of frequency amount",fontsize=16)
plt.show()
```
